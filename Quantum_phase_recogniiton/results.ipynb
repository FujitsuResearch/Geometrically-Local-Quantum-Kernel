```python
import numpy as np
import math
import random as rand
import statistics
import matplotlib.pyplot as plt
from tqdm import tqdm
```

```python
filename = f'results/SK_ndep_nt=250_normalize.npy'
sk250 = np.load(filename)

filename = f'results/SK_ndep_nt=500_normalize.npy'
sk500 = np.load(filename)

filename = f'results/TESK_ndep_nt=250_normalize.npy'
tesk250 = np.load(filename)

filename = f'results/TESK_ndep_nt=500_normalize.npy'
tesk500 = np.load(filename)

plt.figure(figsize=(5,5))
plt.ylim(0.6,1.0)
plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.xticks(fontsize=12)
plt.yticks([0.6,0.7,0.8,0.9,1.0],fontsize=12)
plt.xlabel("Number of qubits", fontsize=16)
plt.ylabel("Test accuracy", fontsize=16)

plt.errorbar(sk500[0,:], sk500[1,:], yerr=sk500[2,:], capsize=2, marker="s", color="tab:orange", label="SK (N=500)")
plt.errorbar(sk250[0,:], sk250[1,:], yerr=sk250[2,:], capsize=2, marker="s", color="tab:orange", linestyle="dashed", label="SK (N=250)")
plt.errorbar(tesk500[0,:], tesk500[1,:], yerr=tesk500[2,:], capsize=2, marker="o", color="tab:blue", label="GLQK (N=500)")
plt.errorbar(tesk250[0,:], tesk250[1,:], yerr=tesk250[2,:], capsize=2, marker="o", color="tab:blue", linestyle="dashed", label="GLQK (N=250)")

plt.legend(fontsize=12)
plt.legend(fontsize=14, handletextpad=0.2, borderpad=0.2, columnspacing=1.0)
plt.savefig(f"figs/ndep_normalize.pdf", bbox_inches="tight", pad_inches=0.1)
```
